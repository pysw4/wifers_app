```python
#Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout
from keras.callbacks import EarlyStopping

import warnings
warnings.filterwarnings('ignore')
```

```python
# ============================================================
#  生成示例数据（因为 meme_clean.parquet 不在本地）
#  模拟 3 个 AP、每个 500 小时、10 个特征的信号数据
# ============================================================
np.random.seed(42)
n_aps = 3
hours_per_ap = 500

records = []
for ap_id in range(n_aps):
    base_signal = 0.5 + 0.3 * np.sin(np.linspace(0, 10*np.pi, hours_per_ap))  # 日周期
    for h in range(hours_per_ap):
        records.append({
            'associated_device_name': f'AP_{ap_id+1}',
            'timestamp': pd.Timestamp('2025-01-01') + pd.Timedelta(hours=h + ap_id * hours_per_ap),
            'signal_score': np.clip(base_signal[h] + np.random.normal(0, 0.05), 0, 1),
            'signal_strength': np.random.randint(-80, -30),
            'signal_db': np.random.uniform(-90, -20),
            'snr': np.random.uniform(10, 40),
            'cpu_utilization': np.random.uniform(10, 90),
            'mem_usage': np.random.uniform(20, 95),
            'client_count': np.random.randint(0, 50),
            'health': np.random.uniform(30, 100),
            'speed': np.random.uniform(50, 1000),
            'maxspeed': np.random.uniform(100, 2000),
        })

df_aps = pd.DataFrame(records)
print(f'Dataset shape: {df_aps.shape}')
df_aps.head()
```

```python
#for now, drop nan
df_aps = df_aps.dropna()
print(f'After dropna: {df_aps.shape}')
```

```python
# Convert timestamp
df_aps['timestamp'] = pd.to_datetime(df_aps['timestamp'])

# Sort values
df_aps_sorted = df_aps.sort_values(['associated_device_name', 'timestamp'])

print(df_aps[['associated_device_name', 'timestamp']].head())
```

```python
#Numerical features
features = ['signal_score', 'signal_strength', 'signal_db', 'snr', 'cpu_utilization', 'mem_usage',
    'client_count', 'health', 'speed', 'maxspeed']

#Relation between name and time stamps
required_columns = ['associated_device_name', 'timestamp'] + features

model_df = df_aps_sorted[required_columns].copy()

model_df.head()
```

```python
# Check missing values
print(model_df.isnull().sum())

# Forward fill per AP
model_df = model_df.groupby('associated_device_name').apply(
    lambda x: x.ffill().bfill()
).reset_index(drop=True)

print(model_df.isnull().sum())
```

```python
#Group the data by hours so we can predict based on hours
hourly_df = (model_df.set_index('timestamp').groupby('associated_device_name').resample('1H').mean(numeric_only=True).reset_index())

hourly_df.head()
```

```python
#Number of hours of data of each ap after the transformations
ap_counts = (hourly_df['associated_device_name'].value_counts())

print(ap_counts.head)
```

```python
# Select one AP
ap_name = hourly_df['associated_device_name'].iloc[0]

ap_df = hourly_df[hourly_df['associated_device_name'] == ap_name].copy()

print(f'Selected AP: {ap_name}')
print(ap_df.shape)

ap_df
```

```python
plt.figure(figsize=(15,5))
plt.plot(ap_df['timestamp'], ap_df['signal_score'])
plt.title(f'Signal Score Over Time — {ap_name}')
plt.xlabel('Time')
plt.ylabel('Signal Score')
plt.grid(True)
plt.show()
```

# Train a nn

```python
#Neural networks work better with normalized values.

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(ap_df[features])

scaled_df = pd.DataFrame(scaled_data, columns=features)

scaled_df.head()
```

We will create a predictor that, based on the past 24 hours, predicts the next 5. So, we create:

*   Input window = previous 24 hours
*   Output horizon = next 5 hours

In other words:

* X → last 24 hours
* y → next 5 values of signal_score

```python
WINDOW_SIZE = 24
FORECAST_HORIZON = 5

TARGET_COLUMN = 'signal_score'

target_idx = features.index(TARGET_COLUMN)


def create_sequences(data, window_size, forecast_horizon, target_idx):
    X = []
    y = []


    for i in range(len(data) - window_size - forecast_horizon):
      X.append(data[i:i+window_size])
      y.append(data[i+window_size:i+window_size+forecast_horizon, target_idx])

    return np.array(X), np.array(y)


X, y = create_sequences(scaled_data, WINDOW_SIZE, FORECAST_HORIZON, target_idx)

print('X shape:', X.shape)
print('y shape:', y.shape)
```

```python
#Check if the the X has samples
print(ap_df.shape)

print(len(scaled_data))

print(len(scaled_data) - WINDOW_SIZE - FORECAST_HORIZON)
```

##Training

```python
#Split into train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print(X_train.shape)
print(X_test.shape)
```

```python
model = Sequential([LSTM(64, return_sequences=True, input_shape=(WINDOW_SIZE, len(features))),

    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(FORECAST_HORIZON)])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()
```

```python
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = model.fit(X_train, y_train, validation_split=0.2, epochs=50, batch_size=32, callbacks=[early_stop])
```

##Evaluation

```python
results = model.evaluate(X_test, y_test)

print(f'Test Loss: {results[0]}')
print(f'Test MAE: {results[1]}')
```

###Prediction

```python
predictions = model.predict(X_test)

print(predictions.shape)
```

```python
#We compare the prediction for the next hour.
plt.figure(figsize=(15,5))

plt.plot(
    y_test[:,0],
    label='Real'
)

plt.plot(
    predictions[:,0],
    label='Predicted'
)

plt.title('Next-Hour Signal Score Prediction')
plt.xlabel('Samples')
plt.ylabel('Signal Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#We use the latest available window to predict the next 5 hours
last_window = scaled_data[-WINDOW_SIZE:]

last_window = np.expand_dims(last_window, axis=0)

future_prediction = model.predict(last_window)

future_prediction = future_prediction[0]

print('Predicted Signal Score for the Next 5 Hours:')

for i, value in enumerate(future_prediction, start=1):
    print(f'Hour +{i}: {value:.4f}')
```

```python
#Convert Predicted Scores into Signal Integrity Labels
def score2integrity(score):

    if score >= 0.95:
        return 'Excellent++'

    elif score >= 0.90:
        return 'Excellent+'

    elif score >= 0.80:
        return 'Excellent'

    elif score >= 0.70:
        return 'Good'

    elif score >= 0.50:
        return 'Fair'

    else:
        return 'Poor'

future_labels = [score2integrity(x) for x in future_prediction]

for i, label in enumerate(future_labels, start=1):
    print(f'Hour +{i}: {label}')
```

## 🚀 扩展：使用全部数据训练滑窗 LSTM，预测未来任意小时

以下新增 cell **不修改**原有 24→5 的代码。核心思想：
- 使用 **滑动窗口** 覆盖全部历史数据，生成大量训练样本
- **WINDOW_SIZE_ALL** 可自由设置（如 24, 72, 168 小时）
- **FH** 可设置为任意值（如 5, 12, 24 小时）
- 预测时用最新 WINDOW_SIZE_ALL 小时的数据，预测未来 FH 小时

```python
# ============================================================
# 配置参数 — 可自由修改
# ============================================================
WINDOW_SIZE_ALL = 72          # 使用最近 72 小时预测（可改为 24, 168 等）
FH = 12                       # 预测未来 12 小时（可改为任意值）

print(f'Window size: {WINDOW_SIZE_ALL} hours')
print(f'Forecast horizon: {FH} hours')
print(f'Total training samples: {len(scaled_data) - WINDOW_SIZE_ALL - FH}')
```

```python
# ============================================================
# 用滑窗在全部数据上生成序列
# ============================================================
X_full, y_full = create_sequences(scaled_data, WINDOW_SIZE_ALL, FH, target_idx)

print(f'X shape: {X_full.shape} — (samples, window_size, features)')
print(f'y shape: {y_full.shape} — (samples, forecast_horizon)')
print(f'数据利用率: {X_full.shape[0]} / {len(scaled_data)} 个可能的起始点 = {100*X_full.shape[0]/len(scaled_data):.1f}%')
```

```python
# ============================================================
# 训练新模型
# ============================================================
Xf_train, Xf_test, yf_train, yf_test = train_test_split(X_full, y_full, test_size=0.2, shuffle=False)

print(f'Train: {Xf_train.shape}')
print(f'Test:  {Xf_test.shape}')
```

```python
model_full = Sequential([
    LSTM(64, return_sequences=True, input_shape=(WINDOW_SIZE_ALL, len(features))),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(FH)   # 输出未来 FH 小时的 signal_score
])

model_full.compile(optimizer='adam', loss='mse', metrics=['mae'])
model_full.summary()
```

```python
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history_full = model_full.fit(
    Xf_train, yf_train,
    validation_split=0.2,
    epochs=50,
    batch_size=32,
    callbacks=[early_stop]
)
```

```python
# ============================================================
# 评估
# ============================================================
results_full = model_full.evaluate(Xf_test, yf_test)
print(f'Test Loss (MSE): {results_full[0]:.6f}')
print(f'Test MAE:        {results_full[1]:.6f}')
```

```python
# ============================================================
# 通用预测函数 — 用最近 WINDOW_SIZE_ALL 小时预测未来任意小时
# ============================================================
def predict_any_future(model, scaled_data, scaler, window_size, future_hours, target_idx=7):
    """
    使用最近 `window_size` 小时数据预测未来 `future_hours` 小时的 signal_score。
    
    Parameters
    ----------
    model : trained Keras model
    scaled_data : np.ndarray, shape (total_hours, n_features)
    scaler : MinMaxScaler fitted on original data
    window_size : int — 输入窗口大小（应与训练时一致）
    future_hours : int — 想要预测的未来小时数（<= 模型训练的 FH）
    target_idx : int — signal_score 在 features 中的索引
    
    Returns
    -------
    predictions : np.ndarray, shape (future_hours,) — 原始尺度的预测值
    """
    # 取最近 window_size 条数据
    last_window = scaled_data[-window_size:]
    
    # 输入形状: (1, window_size, n_features)
    input_seq = np.expand_dims(last_window, axis=0)
    
    # 模型预测
    pred_scaled = model.predict(input_seq, verbose=0)[0]
    
    # 截取需要的长度
    horizon = min(future_hours, len(pred_scaled))
    pred_scaled = pred_scaled[:horizon]
    
    # 如果不够，用最后一个值填充
    if horizon < future_hours:
        last_val = pred_scaled[-1]
        pad = np.full(future_hours - horizon, last_val)
        pred_scaled = np.concatenate([pred_scaled, pad])
    
    # 反标准化
    dummy = np.zeros((future_hours, scaled_data.shape[1]))
    dummy[:, target_idx] = pred_scaled
    pred_original = scaler.inverse_transform(dummy)[:, target_idx]
    
    return pred_original


# 演示：用最近数据预测未来不同长度
print(f'🔮 基于最近 {WINDOW_SIZE_ALL}h 数据，预测未来信号强度：')
for fh_demo in [5, 12, 24]:
    preds = predict_any_future(model_full, scaled_data, scaler, WINDOW_SIZE_ALL, fh_demo, target_idx)
    labels = [score2integrity(s) for s in preds]
    print(f'\n--- 未来 {fh_demo} 小时 ---')
    for i, (val, lbl) in enumerate(zip(preds, labels), start=1):
        print(f'  Hour +{i:>2d}: {val:.4f}  →  {lbl}')
```

```python
# ============================================================
# 测试集预测 vs 真实值对比
# ============================================================
preds_full = model_full.predict(Xf_test)

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

for step in range(4):
    ax = axes[step // 2, step % 2]
    ax.plot(yf_test[:, step], label=f'Real (t+{step+1})', alpha=0.8)
    ax.plot(preds_full[:, step], label=f'Predicted (t+{step+1})', alpha=0.8, linestyle='--')
    ax.set_title(f'Hour +{step+1} Signal Score — Window={WINDOW_SIZE_ALL}h → FH={FH}h')
    ax.set_xlabel('Test Sample')
    ax.set_ylabel('Signal Score (scaled)')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 效果对比
# ============================================================
print('=' * 60)
print(f'  📊 模型效果对比')
print('=' * 60)
print(f'  原模型   (24h→5h):        MSE={results[0]:.6f}, MAE={results[1]:.6f}')
print(f'  全历史滑窗 ({WINDOW_SIZE_ALL}h→{FH}h): MSE={results_full[0]:.6f}, MAE={results_full[1]:.6f}')
print('=' * 60)
print()
print('💡 提示：')
print('  - 修改 WINDOW_SIZE_ALL 可调整输入窗口大小')
print('  - 修改 FH 可预测未来任意小时数')
print('  - 滑窗策略利用了全部历史数据生成训练样本')
```
